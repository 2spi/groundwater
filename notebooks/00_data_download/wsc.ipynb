```python
import pandas as pd
```

```python
station_code = '05BB001'
start_year = '2000'
end_year = '2025'
```

```python
link =  f"https://wateroffice.ec.gc.ca/services/daily_data/csv/inline?stations[]={station_code}&parameters[]=flow&parameters[]=level&start_date={start_year}-01-01&end_date={end_year}-12-31"
```

```python
df = pd.read_csv(link, parse_dates=[0], usecols=[1,2,3], index_col=0)
```

```python
groups = dict(list(df.groupby('Parameter/Paramètre')))
df_flow = groups['discharge/débit']
df_flow = df_flow['Value/Valeur']
df_level = groups['water level/niveau']
df_level = df_level['Value/Valeur']
```
